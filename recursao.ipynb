# Questão 1
Soma de vendas: escreva soma(lista) recursiva que some uma lista de vendas diárias, sem usar sum nem laço. Identifique o caso base e o caso recursivo antes de codar.

```python
def soma_recursiva(lista):
    if lista==[]:
        return 0
    return lista[0] + soma_recursiva(lista[1:])
    
```

```python
soma_recursiva([1,2,3])
```

```python
def soma_steps(lista): #não otimizada para fins de aprendizado
    print("descendo:", lista)
    if not lista:                                   # caso base
        print("caso base: devolvo 0")
        return 0
    resto = soma(lista[1:])                         # a chamada pausa aqui até devolver
    resultado = lista[0] + resto
    print(f"voltando: {lista[0]} + {resto} = {resultado}")
    return resultado

```

```python
lista = [1,3,4,5,6,7,10,-1]

soma_steps(lista)
```

# Questão 2
2. Inverter e contar: implemente contar_ocorrencias(lista, valor) recursiva. Depois, rode no Python Tutor e desenhe a pilha de chamadas para uma lista de 4 elementos.


```python
def inverter_contar(lista,valor):
    #print('lista em verificação: ', lista)
    if not lista:
        return 0
    encontrou =  1 if lista[0] == valor else 0
    return encontrou + inverter_contar(lista[1:], valor)
```

```python
inverter_contar((1,1,1,1,1,1,2,2,2,3,4,5,6,7,8,9,0),2)
```

```python
def contar_ocorrencias(lista, valor):
    if not lista:                                   # caso base
        print("caso base: lista vazia, sem elemento para guardar. Devolvo 0")
        return 0
    elemento = lista[0]                             # o que este nível guarda
    achou = 1 if elemento == valor else 0
    print(f"descendo: guardo elemento={elemento} (achou={achou}) | lista={lista}")
    resto = contar_ocorrencias(lista[1:], valor)    # a chamada pausa aqui até devolver
    resultado = achou + resto
    print(f"voltando: elemento={elemento} (achou={achou}) + resto={resto} = {resultado}")
    return resultado

print("resultado final:", contar_ocorrencias([1,1,1,4,5,1,7,8,1,0], 1))
```

# Questão 3: Suavização exponencial: 

implemente S_t = α·x_t + (1-α)·S_(t-1) de forma recursiva sobre uma série. Compare o resultado com pandas.ewm(adjust=False) e explique por que a versão recursiva se torna um problema em séries longas.



```python
def suavizacao_exp(serie, alpha):
    if not serie:
        return []
    if len(serie) == 1:
        return [serie[0]] # cria uma lista nova que contem o elemento selecionado
    anterior = suavizacao_exp(serie[:-1], alpha)
    novo = alpha * serie[-1] + (1- alpha) * anterior[-1]
    return anterior + [novo]
```

```python
suavizacao_exp([10,12,9], 0.5)
```

```python
import pandas as pd


def suavizacao_exponencial(serie, alpha, nivel=0):
    recuo = "    " * nivel
    rotulo = f"{recuo}[nível {nivel}]"

    print(f"{rotulo} DESCENDO | recebi serie={serie}")

    if not serie:                                       # caso base: série vazia
        print(f"{rotulo} caso base (série vazia) | devolvo []")
        return []

    if len(serie) == 1:                                 # caso base: S_0 = x_0
        print(f"{rotulo} caso base (1 elemento) | S_0 = x_0 = {serie[0]} | devolvo {[serie[0]]}")
        return [serie[0]]

    x_t = serie[-1]                                     # observação do ponto atual
    t = len(serie) - 1                                  # posição deste ponto na série
    print(f"{rotulo} este nível cuida do ponto t={t} | x_{t} = {x_t}")
    print(f"{rotulo} para calcular S_{t} preciso de S_{t - 1}, então peço a lista de {serie[:-1]}")

    anterior = suavizacao_exponencial(serie[:-1], alpha, nivel + 1)

    s_anterior = anterior[-1]                           # S_(t-1), devolvido pela chamada de baixo
    parte_nova = alpha * x_t
    parte_antiga = (1 - alpha) * s_anterior
    novo = parte_nova + parte_antiga
    resultado = anterior + [novo]

    print(f"{rotulo} VOLTANDO | a chamada de baixo devolveu {anterior}")
    print(f"{rotulo}   S_(t-1) = S_{t - 1} = {s_anterior}  (último valor da lista que voltou)")
    print(f"{rotulo}   S_{t} = alfa*x_{t} + (1-alfa)*S_{t - 1}")
    print(f"{rotulo}   S_{t} = {alpha}*{x_t} + {1 - alpha}*{s_anterior} = {parte_nova} + {parte_antiga} = {novo}")
    print(f"{rotulo}   devolvo {anterior} + [{novo}] = {resultado}")
    return resultado


```

```python
serie = [10, 12, 9]
alpha = 0.3

obtido = suavizacao_exponencial(serie, alpha)

esperado = pd.Series(serie).ewm(alpha=alpha, adjust=False).mean().tolist()
print()
print("resultado da função:", obtido)
print("pandas ewm         :", esperado)
print("bate com o pandas  :", all(abs(a - b) < 1e-9 for a, b in zip(obtido, esperado)))
```

##### Análise de complexidade da versão recursiva

Avalio a função `suavizacao_exponencial` sob três ângulos: tempo, memória e profundidade da recursão.

**Tempo: $O(n^2)$.** A função faz $n$ chamadas, e a conta da suavização em cada uma custa $O(1)$. O custo vem das cópias de lista: o fatiamento `serie[:-1]` na descida e a concatenação `anterior + [novo]` na volta. Como o tamanho dessas listas diminui a cada nível, a soma das cópias é $(n-1) + (n-2) + \dots + 1 = \frac{n(n-1)}{2}$, ou seja, $O(n^2)$. **O termo quadrático vem da soma de cópias de tamanho decrescente, e não de percorrer a série inteira a cada chamada.**

**Memória: $O(n^2)$.** Cada nível guarda a sua fatia enquanto aguarda a chamada seguinte. Dessa forma, no ponto mais fundo, coexistem fatias de tamanho $n-1, n-2, \dots, 1$.

**Profundidade: $n$.** A pilha cresce com o tamanho da série. Por esta razão, o limite de recursão do Python (em geral 1000) impede o uso desta versão em séries longas.
